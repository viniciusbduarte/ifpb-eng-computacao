```python
# 2. Explique, com suas palavras, os conceitos de vetorização e broadcasting que Numpy implementa.

# R
# A vetorização implementada pelo NumPy permite realizar operações sobre um conjunto de dados
# (vetores e matrizes) de uma só vez. Dessa forma, é possível, por exemplo, somar um valor a
# todos os elementos de um vetor sem a necessidade de utilizar um laço de repetição.
#
# Já o broadcasting é o mecanismo que permite realizar operações entre arrays de tamanhos ou
# dimensões diferentes. O NumPy expande automaticamente o array menor (ou um valor escalar)
# para que suas dimensões sejam compatíveis com o array maior durante a operação.
```

```python
# 3. Liste as diferenças entre uma lista, um dicionário, um array numpy, dando exemplos deaplicação de em que cada tipo seria mais interessante.

# R
# Lista:
# Uma lista é uma estrutura de dados que armazena elementos em uma sequência ordenada,
# permitindo diferentes tipos de dados (inteiros, strings, objetos etc.) em uma mesma lista.
# É indicada para armazenar coleções de dados gerais, como uma lista de nomes, produtos ou arquivos.

# Dicionário:
# Um dicionário armazena dados em pares chave-valor, permitindo acessar cada informação
# por meio de uma chave, em vez de um índice. É útil para representar objetos ou registros,
# como informações de um aluno (nome, matrícula e curso).

# Array NumPy:
# Um array NumPy é uma estrutura otimizada para armazenar grandes conjuntos de dados
# numéricos de um mesmo tipo. Permite realizar operações matemáticas de forma vetorizada
# e muito mais eficiente do que listas do Python. É amplamente utilizado em Métodos
# Numéricos, álgebra linear, processamento de sinais, inteligência artificial e análise de dados.
```

```python
# 6. Considere a lista [-2, 4., ’7’, 9] e crie um array Numpy a partir dessa lista. 
# Houve uma mudança no tipo de vários elementos. Explique com suas palavras porque isso aconteceu. 
# Se o elemento ’7’ for escrito como inteiro, o que muda na saída e por que?
```

```python
lista = [-2, 4., '7', 9]

print(lista) 

list_arr_numpy = np.array(lista)

print(list_arr_numpy)
print(list_arr_numpy.dtype)

lista2 = [-2, 4., 7, 9]

print(lista2) 

list2_arr_numpy = np.array(lista2)

print(list2_arr_numpy)
print(list2_arr_numpy.dtype)
```

```python
# R
# Ao criar um array NumPy a partir da lista [-2, 4., '7', 9], os elementos são
# convertidos para string. Isso acontece porque um array NumPy deve armazenar elementos
# de um único tipo de dado, e a presença de uma string faz com que os demais elementos
# sejam convertidos para esse tipo.
#
# Se o elemento '7' for escrito como inteiro, o array passará a conter apenas números
# inteiros e reais. Nesse caso, o NumPy converterá todos os elementos para float, pois esse
# tipo consegue representar tanto valores inteiros quanto decimais sem perder informação.
```

```python
# 7. Crie um vetor A com valores variando de 10 à 49, um vetor B variando de 1 à 39 e um vetor C de 10 à 20. 
# Então realize e comente a saída das seguintes operações:

# (a) A+ B
# (b) A· B
# (c) A+ C
# (d) B · C
```

```python
A = np.random.randint(10, 49, 10)
B = np.random.randint(1, 39, 10)
C = np.random.randint(10, 20, 10)

print("A = ", A, " B = ", B, " C = ", C)

print("A + B = ", A + B)
print("A * B = ", A * B)
print("A + C = ", A + C)
print("B * C = ", B * C)
```

```python
# 10. Usando a função sin do numpy, e a função plot da matplotlib, módulo de geração de gráficos, plote um gráfico contendo os dois primeiros períodos de uma senóide.
```

```python
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("Função Seno")
plt.xlabel("x")
plt.ylabel("sen(x)")
plt.grid(True)
plt.show()
```
