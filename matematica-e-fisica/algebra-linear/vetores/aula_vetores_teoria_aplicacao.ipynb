## 1. O que é um vetor?

### Teoria

Em álgebra linear, um vetor é uma lista ordenada de números:

$$\vec{v} = [v_1, v_2, \dots, v_N]$$

O número de elementos determina a sua **dimensionalidade** matemática ($v \in \mathbb{R}^N$).

* $\vec{v} = [3, 4]$ é um vetor em $\mathbb{R}^2$ (dimensão 2).
* $\vec{u} = [2, 5, 1]$ é um vetor em $\mathbb{R}^3$ (dimensão 3).

### Aplicação

Em ciência da computação e robótica, vetores são frequentemente usados para representar posições, estados de sensores, forças ou atributos de dados (features).

```python
import numpy as np

v = np.array([3, 4])
u = np.array([2, 5, 1])

print("v =", v)
print("u =", u)
print("shape de v =", v.shape)
print("shape de u =", u.shape)
```

## 2. Vetor como seta (Interpretação Geométrica)

### Teoria

Geometricamente, um vetor é definido por uma seta que possui **magnitude** (comprimento), **direção** (inclinação) e **sentido**.

Quando a cauda do vetor coincide com a origem $(0, 0)$, dizemos que ele está na **posição padrão** (*standard position*). Nesse caso, as coordenadas da ponta da seta coincidem com os elementos do vetor:

$$\vec{v} = [3, 4]$$

### Aplicação

Se um robô está na posição inicial $(0, 0)$ e recebe um vetor de deslocamento $[3, 4]$, a sua nova coordenada será dada pela soma direta dos vetores.

```python
posicao = np.array([0, 0])
deslocamento = np.array([3, 4])

print("Nova posição:", posicao + deslocamento)
```

## 3. Adição de Vetores

### Teoria

A soma vetorial só é permitida entre vetores de mesma dimensionalidade e ocorre elemento a elemento:

$$\vec{a} + \vec{b} = [a_1 + b_1, \; a_2 + b_2, \; \dots, \; a_N + b_N]$$

### Aplicação

Composição de dois deslocamentos sequenciais:

$$[3, 2] + [1, 4] = [4, 6]$$

```python
a = np.array([3, 2])
b = np.array([1, 4])

print("a + b =", a + b)
```

## 4. Subtração de Vetores

### Teoria

Assim como a soma, a subtração ocorre elemento por elemento:

$$\vec{a} - \vec{b} = [a_1 - b_1, \; a_2 - b_2, \; \dots, \; a_N - b_N]$$

Geometricamente, $\vec{b} - \vec{a}$ representa o vetor que aponta da cabeça de $\vec{a}$ para a cabeça de $\vec{b}$.

### Aplicação

Para encontrar o deslocamento relativo entre dois pontos $A = [2, 3]$ e $B = [5, 7]$:

$$\vec{AB} = B - A = [5 - 2, \; 7 - 3] = [3, 4]$$

```python
A = np.array([2, 3])
B = np.array([5, 7])

print("Vetor AB:", B - A)
```

## 5. Multiplicação por Escalar

### Teoria

Um escalar $\lambda$ redimensiona (*scale*) o vetor, multiplicando cada componente individualmente:

$$\lambda \vec{v} = [\lambda v_1, \; \lambda v_2, \; \dots, \; \lambda v_N]$$

* Se $\lambda > 1$: o vetor é ampliado.
* Se $0 < \lambda < 1$: o vetor é contraído.
* Se $\lambda < 0$: o vetor inverte de sentido em $180^\circ$.

### Aplicação

$$3 \cdot [2, 3] = [6, 9]$$

```python
v = np.array([2, 3])
print("3 * v =", 3 * v)
```

## 6. Vetor Nulo

### Teoria

O vetor nulo (representado por $\mathbf{0}$) possui todos os seus elementos iguais a zero:

$$\mathbf{0} = [0, 0, \dots, 0]$$

Ele atua como elemento neutro na adição vetorial:

$$\vec{v} + \mathbf{0} = \vec{v}$$

### Aplicação

Adicionar um vetor de velocidade ou deslocamento nulo mantém o estado invariante.

```python
v = np.array([5, 3])
zero = np.array([0, 0])

print("v + zero =", v + zero)
```

## 7. Norma (Magnitude) de um Vetor

### Teoria

A norma Euclidiana (ou comprimento geométrico) de um vetor $\vec{v}$ é dada por:

$$\Vert{}\vec{v}\Vert{} = \sqrt{\sum_{i=1}^{N} v_i^2}$$

Para um vetor bidimensional:

$$\Vert{}\vec{v}\Vert{} = \sqrt{x^2 + y^2}$$

### Aplicação

Para $\vec{v} = [3, 4]$:

$$\Vert{}\vec{v}\Vert{} = \sqrt{3^2 + 4^2} = \sqrt{9 + 16} = 5$$

```python
v = np.array([3, 4])
print("Norma:", np.linalg.norm(v))
```

## 8. Vetor Unitário (Normalização)

### Teoria

Um vetor unitário tem comprimento exatamente igual a 1. É calculado multiplicando o vetor pelo inverso da sua norma:

$$\hat{v} = \frac{\vec{v}}{\Vert{}\vec{v}\Vert{}}, \qquad \Vert{}\hat{v}\Vert{} = 1$$

*(O vetor nulo $\mathbf{0}$ é o único vetor que não possui vetor unitário associado, pois $\Vert{}\mathbf{0}\Vert{} = 0$.)*

### Aplicação

Para $\vec{v} = [3, 4]$:

$$\hat{v} = \frac{[3, 4]}{5} = [0.6, 0.8]$$

Isso isola a informação de direção, descartando a escala de grandeza.

```python
v = np.array([3.0, 4.0])
unitario = v / np.linalg.norm(v)

print("Unitário:", unitario)
print("Norma do unitário:", np.linalg.norm(unitario))
```

## 9. Produto Escalar (Dot Product)

### Teoria

O produto escalar entre dois vetores de mesma dimensionalidade é obtido multiplicando os elementos correspondentes e somando os produtos parciais:

$$\vec{a} \cdot \vec{b} = a^T b = \sum_{i=1}^N a_i b_i$$

O resultado final é sempre um número **escalar**.

### Aplicação

$$[1, 2, 3] \cdot [4, 5, 6] = (1 \times 4) + (2 \times 5) + (3 \times 6) = 4 + 10 + 18 = 32$$

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Produto escalar:", np.dot(a, b))
```

## 10. Interpretação Geométrica e Ortogonalidade

### Teoria

Geometricamente, o produto escalar relaciona o comprimento dos vetores e o ângulo $\theta$ entre eles:

$$\vec{a} \cdot \vec{b} = \Vert{}\vec{a}\Vert{} \Vert{}\vec{b}\Vert{} \cos(\theta)$$

O sinal do produto escalar indica a orientação mútua:
* $\theta < 90^\circ \implies \vec{a} \cdot \vec{b} > 0$ (ângulo agudo)
* $\theta = 90^\circ \implies \vec{a} \cdot \vec{b} = 0$ (**vetores ortogonais**)
* $\theta > 90^\circ \implies \vec{a} \cdot \vec{b} < 0$ (ângulo obtuso)

### Aplicação

Para os eixos cartesianos $[1, 0]$ e $[0, 1]$:

$$[1, 0] \cdot [0, 1] = (1 \times 0) + (0 \times 1) = 0$$

Logo, eles formam um ângulo de $90^\circ$ e são ortogonais.

```python
a = np.array([1, 0])
b = np.array([0, 1])

produto = np.dot(a, b)
print("Produto escalar:", produto)
print("Ortogonais?", np.isclose(produto, 0.0))
```

## 11. Multiplicação Elemento a Elemento (Hadamard)

### Teoria

O produto de Hadamard multiplica elementos na mesma posição, resultando em um **novo vetor** de mesma dimensionalidade:

$$\vec{a} \odot \vec{b} = [a_1 b_1, \; a_2 b_2, \; \dots, \; a_N b_N]$$

No NumPy, o operador `*` executa o produto de Hadamard (diferente de `np.dot` que faz o produto escalar).

### Aplicação

$$[1, 2, 3] \odot [4, 5, 6] = [4, 10, 18]$$

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Hadamard (a * b)   :", a * b)
print("Escalar np.dot(a, b):", np.dot(a, b))
```

## 12. Projeção e Decomposição Ortogonal

### Teoria

A projeção ortogonal do vetor $\vec{a}$ sobre a reta de referência definida por $\vec{b}$ é o componente paralelo a $\vec{b}$:

$$\vec{a}_{\parallel b} = \left( \frac{\vec{a} \cdot \vec{b}}{\vec{b} \cdot \vec{b}} \right) \vec{b} = \left( \frac{\vec{a} \cdot \vec{b}}{\Vert{}\vec{b}\Vert{}^2} \right) \vec{b}$$

O componente perpendicular (resíduo ortogonal) é obtido por subtração:

$$\vec{a}_{\perp b} = \vec{a} - \vec{a}_{\parallel b}$$

Permitindo decompor $\vec{a}$ em dois vetores mutuamente perpendiculares:

$$\vec{a} = \vec{a}_{\parallel b} + \vec{a}_{\perp b}$$

### Aplicação

Se $\vec{a} = [3, 2]$ e $\vec{b} = [1, 0]$:

$$\vec{a}_{\parallel b} = [3, 0]$$
$$\vec{a}_{\perp b} = [3, 2] - [3, 0] = [0, 2]$$

Reconstrução:

$$[3, 2] = [3, 0] + [0, 2]$$

```python
a = np.array([3.0, 2.0])
b = np.array([1.0, 0.0])

projecao = (np.dot(a, b) / np.dot(b, b)) * b
perpendicular = a - projecao

print("Paralela     :", projecao)
print("Perpendicular:", perpendicular)
print("Reconstrução :", projecao + perpendicular)
```
